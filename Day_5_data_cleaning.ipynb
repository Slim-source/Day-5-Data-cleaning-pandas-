```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data=pd.read_csv("student_dropout_risk_2026.csv")
print(data.shape)
print(data.head())
```

```python
print(data.isnull().sum())
print(data.dtypes)
```

```python
lvisit_mode=data['library_visits_per_month'].mode()[0]
data['library_visits_per_month']=data['library_visits_per_month'].fillna(lvisit_mode)

lmslogins_mean=data['lms_logins_per_week'].mean()
data['lms_logins_per_week']=data['lms_logins_per_week'].fillna(lmslogins_mean)

ecurricular_mode=data['extracurricular_activities'].mode()[0]
data['extracurricular_activities']=data['extracurricular_activities'].fillna(ecurricular_mode)

mdeadline_mode=data['missed_deadlines'].mode()[0]
data['missed_deadlines']=data['missed_deadlines'].fillna(mdeadline_mode)

stresslevel_mean=data['stress_level'].mean()
data['stress_level']=data['stress_level'].fillna(stresslevel_mean)

data['mental_health_support']=data['mental_health_support'].fillna(0.0)

satisfaction_mode=data['satisfaction_score'].mode()[0]
data['satisfaction_score']=data['satisfaction_score'].fillna(satisfaction_mode)

commute_mean=data['commute_time_mins'].mean()
data['commute_time_mins']=data['commute_time_mins'].fillna(commute_mean)

sleephrs_mean=data['sleep_hours'].mean()
data['sleep_hours']=data['sleep_hours'].fillna(sleephrs_mean)

data['dropout']=data['dropout'].fillna(0.0)



print(data.isnull().sum())
```
